```python
import requests
import pandas as pd
import time
import random
from fake_useragent import UserAgent
import os

# ==============================
# CONFIG
# ==============================
ua = UserAgent()

# Proxy list
proxies_auth = [
    "http://ce42509537a8eed9520a:25ea7a1cfec42236@gw.dataimpulse.com:10000",
    "http://ce42509537a8eed9520a:25ea7a1cfec42236@gw.dataimpulse.com:10001",
    "http://ce42509537a8eed9520a:25ea7a1cfec42236@gw.dataimpulse.com:10002",
    "http://ce42509537a8eed9520a:25ea7a1cfec42236@gw.dataimpulse.com:10003",
    "http://ce42509537a8eed9520a:25ea7a1cfec42236@gw.dataimpulse.com:10004",
    "http://ce42509537a8eed9520a:25ea7a1cfec42236@gw.dataimpulse.com:10005",
    "http://ce42509537a8eed9520a:25ea7a1cfec42236@gw.dataimpulse.com:10006",
    "http://ce42509537a8eed9520a:25ea7a1cfec42236@gw.dataimpulse.com:10007",
    "http://ce42509537a8eed9520a:25ea7a1cfec42236@gw.dataimpulse.com:10008",
    "http://ce42509537a8eed9520a:25ea7a1cfec42236@gw.dataimpulse.com:10009"
]

# Load CSV with URLs
df = pd.read_csv("C:\\Users\\MICILMEDS\\Documents\\Medi_final\\correct_data\\Category\\pain_relief.csv")

# Directory to save HTML files
os.makedirs("html_pages", exist_ok=True)

# ==============================
# SCRAPER
# ==============================
for num in range(len(df)):
    url = df.iloc[num, 5]

    # Pick random proxy & user-agent
    proxy = random.choice(proxies_auth)
    proxies = {"http": proxy, "https": proxy}
    headers = {
        "Connection": "keep-alive",
        "User-Agent": ua.random,
        "Accept-Language": "en-US,en;q=0.9"
    }

    try:
        res = requests.get(url, headers=headers, proxies=proxies, timeout=20)

        if res.status_code == 404:
            print(f"[{num}] ❌ 404 Not Found: {url}")
            continue

        # Save raw HTML into a file
        filename = f"html_pages/page_{num}.html"
        with open(filename, "w", encoding="utf-8") as f:
            f.write(res.text)

        print(f"[{num}] ✅ Saved HTML ({len(res.text)} chars) | Proxy: {proxy.split('@')[-1]} | Status: {res.status_code}")

    except Exception as e:
        print(f"[{num}] ⚠️ Error with proxy {proxy}: {e}")
        time.sleep(5)
        continue

    # Anti-blocking sleep
    if num % 50 == 0 and num > 0:
        time.sleep(random.uniform(8, 15))
    else:
        time.sleep(random.uniform(2, 6))
    if num ==5:
        break

print("✅ All HTML pages downloaded.")
```
