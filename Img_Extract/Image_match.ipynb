```python
import os

df_test['Product Name_x'] = df_test['Product Name_x'].str.replace(" ", "_")

image_folder = r"downloaded_products\Abbot"
ext = ".jpg"

for _, row in df_test.iterrows():
    old_name = os.path.join(image_folder, row["Product Name_x"] + ext)
    new_name = os.path.join(image_folder, row["Product Code"] + ext)

    if os.path.exists(old_name):
        if not os.path.exists(new_name):   # only rename if target does not exist
            os.rename(old_name, new_name)
            print(f"Renamed: {row['Product Name_x']}{ext} -> {row['Product Code']}{ext}")
        else:
            print(f"Skipped (already exists): {new_name}")
    else:
        print(f"File not found: {old_name}")
```

```python
import os
from PIL import Image

# Path to your image folder
image_folder = r"downloaded_products\Abbot"

# Target size
target_size = (1024, 871)

for filename in os.listdir(image_folder):
    if filename.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".tiff", ".webp")):
        file_path = os.path.join(image_folder, filename)
        try:
            with Image.open(file_path) as img:
                # Keep aspect ratio
                img.thumbnail(target_size, Image.LANCZOS)

                # Create a blank canvas (white background)
                new_img = Image.new("RGB", target_size, (255, 255, 255))

                # Center the resized image on the canvas
                x_offset = (target_size[0] - img.width) // 2
                y_offset = (target_size[1] - img.height) // 2
                new_img.paste(img, (x_offset, y_offset))

                # Overwrite the original image
                new_img.save(file_path)
                print(f"Resized (in place): {filename}")
        except Exception as e:
            print(f"Error with {filename}: {e}")
```

```python
import os
import re

# 🔹 Change this to your folder path
folder_path = r"downloaded_products\Abbot"

# Regex pattern: EMD followed by exactly 6 digits
pattern = re.compile(r"^(EMD\d{6})(\..+)?$")

for filename in os.listdir(folder_path):
    match = pattern.match(filename)
    if match:
        base_name = match.group(1)   # EMDXXXXXX
        extension = match.group(2) if match.group(2) else ""  # keep extension if exists
        old_path = os.path.join(folder_path, filename)
        new_name = f"{base_name}_1{extension}"
        new_path = os.path.join(folder_path, new_name)

        # Rename only if not already renamed
        if not os.path.exists(new_path):
            os.rename(old_path, new_path)
            print(f"Renamed: {filename} → {new_name}")
        else:
            print(f"Skipped (already exists): {new_name}")
```
